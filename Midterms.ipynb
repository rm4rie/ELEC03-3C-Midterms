<h2>Extracting Date, Month, Hour & DayOfWeek</h2>

```python
df['Date'] = pd.to_datetime(df['Date'])

df['Month'] = df['Date'].dt.month
df['Hour'] = df['Date'].dt.hour
df['DayOfWeek'] = df['Date'].dt.day_name()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

<H1>Insights</h1>

<h2>1. Are violent crimes (battery, assault) more common near schools compared to other areas?</h2>

```python
school = df[df['Location Description'].str.contains("SCHOOL", na=False)]
other = df[~df['Location Description'].str.contains("SCHOOL", na=False)]

school_rate = (school['Primary Type'].isin(["BATTERY","ASSAULT"])).mean()
other_rate = (other['Primary Type'].isin(["BATTERY","ASSAULT"])).mean()

print("School:", school_rate)
print("Other:", other_rate)
```

<H>Insight: Insight: Crimes near schools are more violent (~64%) compared to other areas (~26%).</h>

<h2> 2. Are gun-related crimes more likely to result in arrest than non-gun crimes? </h2>

```python
gun = df[df['Description'].str.contains("FIREARM", na=False)]
nongun = df[~df['Description'].str.contains("FIREARM", na=False)]

print("Gun arrest rate:", gun['Arrest'].mean())
print("Non-gun arrest rate:", nongun['Arrest'].mean())
```

<h>Insight: Arrest rates differ slightly between gun and non-gun crimes.</h>

<h2> 3.Which FBI crime codes are most linked with arrests, and how do they vary by district?</h2>

```python
fbi_district = (
    df.groupby(["FBI Code", "District"])["Arrest"]
    .agg(["count", "mean"])
    .reset_index()
    .rename(columns={"count": "Total_Crimes", "mean": "Arrest_Rate"})
    .sort_values("Arrest_Rate", ascending=False)
)

fbi_district.head(10)
```

<h>Insight: Shows which categories (as defined by FBI codes) are better enforced by police.</h>

<h2>4. Are narcotics arrests more common in public housing than other areas? </h2>

```python
housing = df[(df['Location Description'].str.contains("PUBLIC HOUSING", na=False)) & (df['Primary Type']=="NARCOTICS")]
other = df[(df['Primary Type']=="NARCOTICS") & (~df['Location Description'].str.contains("PUBLIC HOUSING", na=False))]

print("Housing arrest rate:", housing['Arrest'].mean())
print("Other arrest rate:", other['Arrest'].mean())
```

<h>Insight: Arrest rates for narcotics are higher in public housing.</h>

<h2>5. Are drug crimes in public housing more likely to get arrests than in other places? </h2>

```python
housing = df[(df['Location Description'].str.contains("PUBLIC HOUSING", na=False)) & (df['Primary Type']=="NARCOTICS")]
other = df[(df['Primary Type']=="NARCOTICS") & ~df['Location Description'].str.contains("PUBLIC HOUSING", na=False)]

print("Housing arrest rate:", housing['Arrest'].mean())
print("Other arrest rate:", other['Arrest'].mean())
print("Correlation housing vs other counts:", housing.groupby('Month').size().corr(other.groupby('Month').size()))
```

<h>Insight: Housing drug crimes get more arrests, and both rise and fall together by month.</h>

<h2>6. Which community areas report the most arrests for theft-related crimes?</h2>

```python
theft_areas = (
    df[df["Primary Type"] == "THEFT"]
    .groupby("Community Area")["Arrest"]
    .agg(["count", "mean"])
    .reset_index()
    .rename(columns={"count": "Theft_Cases", "mean": "Arrest_Rate"})
    .sort_values("Theft_Cases", ascending=False)
)
theft_areas.head(10)
```

<h>Insight: The findings identify which neighborhoods are hotspots for theft and whether police usually succeed in making arrests in those areas.</h>

<h2>7. Do robberies in gas stations, groceries, and parking lots happen at different times?</h2>

```python
gas = df[(df['Location Description'].str.contains("GAS STATION", na=False)) & (df['Primary Type']=="ROBBERY")]
gro = df[(df['Location Description'].str.contains("GROCERY", na=False)) & (df['Primary Type']=="ROBBERY")]
park = df[(df['Location Description'].str.contains("PARKING LOT", na=False)) & (df['Primary Type']=="ROBBERY")]

print("Gas avg hour:", gas['Hour'].mean())
print("Grocery avg hour:", gro['Hour'].mean())
print("Parking avg hour:", park['Hour'].mean())

print("Correlation gas vs parking:", gas.groupby('Hour').size().corr(park.groupby('Hour').size()))
```

<h>Insight: Parking lot robberies are later, and gas vs parking follow a similar time trend.</h>

<h2>8. Do robberies on CTA buses and trains happen at different times, and which leads to more arrests?</h2>

```python
bus = df[(df['Location Description'].str.contains("CTA BUS", na=False)) & (df['Primary Type']=="ROBBERY")]
train = df[(df['Location Description'].str.contains("CTA TRAIN", na=False)) & (df['Primary Type']=="ROBBERY")]

print("Bus avg hour:", bus['Hour'].mean())
print("Train avg hour:", train['Hour'].mean())
print("Bus arrest rate:", bus['Arrest'].mean())
print("Train arrest rate:", train['Arrest'].mean())
print("Correlation bus vs train by hour:", bus.groupby('Hour').size().corr(train.groupby('Hour').size()))
```

<h>Insight: Train robberies happen later than bus robberies, and train cases have slightly more arrests.</h>

<h2> 9. Are assaults more frequent on weekends, and do they get more arrests? </h2>

```python
assaults = df[df['Primary Type']=="ASSAULT"]
weekend = assaults[assaults['DayOfWeek'].isin(["Saturday","Sunday"])]
weekday = assaults[~assaults['DayOfWeek'].isin(["Saturday","Sunday"])]

print("Weekend share:", len(weekend)/len(assaults))
print("Weekend arrest rate:", weekend['Arrest'].mean())
print("Weekday arrest rate:", weekday['Arrest'].mean())
print("Correlation arrests vs day:", assaults.groupby('Hour')['Arrest'].mean().corr(pd.Series(range(24))))
```

<h>Insight: Assaults are more common on weekends, but weekday and weekend arrest rates are close.</h>

<h2> 10.Are burglaries in residential buildings more likely to be solved than in commercial buildings?</h2>

```python
res = df[(df['Location Description'].str.contains("RESIDENCE", na=False)) & (df['Primary Type']=="BURGLARY")]
com = df[(df['Location Description'].str.contains("COMMERCIAL", na=False)) & (df['Primary Type']=="BURGLARY")]

print("Residential arrest rate:", res['Arrest'].mean())
print("Commercial arrest rate:", com['Arrest'].mean())
print("Correlation res vs com by month:", res.groupby('Month').size().corr(com.groupby('Month').size()))
```

<h>Insight: Residential burglaries are arrested less, and both trends are linked by month.</h>

<h>Insight: Street assaults are higher, and assault patterns move along with burglary counts.</h>

<h3>1. Crime Counts by Beat and Month</h3>

```python
import pandas as pd
import plotly.express as px
import plotly.io as pio

crime_beat_month = (
    df.groupby(["Beat", "Month"])
    .size()
    .reset_index(name="Crime_Count")
    .pivot(index="Beat", columns="Month", values="Crime_Count")
    .fillna(0)
)

fig = px.imshow(
    crime_beat_month,
    color_continuous_scale="YlOrBr",
    labels=dict(x="Month", y="Beat", color="Crime Count"),
    title="Heatmap of Crime Counts by Beat and Month"
)

fig.show()  # 👈 Display in Jupyter
fig.write_html("heatmap_crime_by_beat_month.html")
```

<h>Insight: This shows crime seasonality at the neighborhood beat level. Some beats may have strong seasonal spikes (e.g., summer crimes), while others remain steady throughout the year.</h>

<h3>2. Crime Description vs Primary Type</h3>

```python
import pandas as pd
import plotly.express as px
import plotly.io as pio

crime_desc_type = (
    df.groupby(["Primary Type", "Description"])
    .size()
    .reset_index(name="Count")
    .pivot(index="Primary Type", columns="Description", values="Count")
    .fillna(0)
)

fig = px.imshow(
    crime_desc_type,
    color_continuous_scale="Blues",  # pinalitan ng ibang kulay
    labels=dict(x="Description", y="Primary Type", color="Count"),
    title="Heatmap of Crime Descriptions by Primary Type"
)

# Styling malinis at malaki
fig.update_layout(
    width=1300,   # lapad ng figure
    height=800,   # taas ng figure
    font=dict(size=14, family="Times New Roman"),
    title=dict(font=dict(size=20)),
    xaxis=dict(
        tickangle=-90,   # nakatagilid labels para readable
        automargin=True,
        tickfont=dict(size=12)
    ),
    yaxis=dict(
        automargin=True,
        tickfont=dict(size=12)
    ),
    coloraxis_colorbar=dict(
        title=dict(
            text="Count",
            font=dict(size=16, family="Times New Roman")
        ),
        tickfont=dict(size=12)
    )
)



fig.update_traces(
    xgap=2, ygap=2
)

fig.show()
fig.write_html("heatmap_crime_desc_vs_type.html")
```

<h>Insight: This heatmap shows the detailed breakdown of what sub-crimes are driving each primary type. For example, theft might mostly be “retail theft,” while assault could be dominated by “simple assault.”</h>
